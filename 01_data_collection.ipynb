# Customer Churn Intelligence Platform
## 01 - Data Collection and Initial Inspection

### Objective
Collect and inspect the original Telco Customer Churn dataset.

### Data source
IBM Telco Customer Churn sample dataset.

### Important
The original dataset must remain unchanged.
All transformations will be performed on copies.

```python
from pathlib import Path

import pandas as pd
import numpy as np

# Racine du projet
PROJECT_ROOT = Path.cwd().parent

# Chemin du dataset
RAW_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "Telco-Customer-Churn.csv"
)

print("Project root:", PROJECT_ROOT)
print("Dataset exists:", RAW_DATA_PATH.exists())
print("Dataset path:", RAW_DATA_PATH)
```

```python
if not RAW_DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {RAW_DATA_PATH}"
    )

df_raw = pd.read_csv(RAW_DATA_PATH)

print("Dataset successfully loaded.")
print(f"Rows: {df_raw.shape[0]}")
print(f"Columns: {df_raw.shape[1]}")

display(df_raw.head())
```

```python
print("DATASET SHAPE")
print(df_raw.shape)

print("\nDATA TYPES")
display(df_raw.dtypes)

print("\nDATASET INFORMATION")
df_raw.info()

print("\nSTATISTICAL SUMMARY")
display(df_raw.describe(include="all").T)
```

```python
quality_report = pd.DataFrame({
    "data_type": df_raw.dtypes.astype(str),
    "missing_values": df_raw.isna().sum(),
    "missing_percentage": (
        df_raw.isna().mean() * 100
    ).round(2),
    "unique_values": df_raw.nunique()
})

display(
    quality_report.sort_values(
        by="missing_percentage",
        ascending=False
    )
)
```

```python
print(
    "Duplicate rows:",
    df_raw.duplicated().sum()
)

print(
    "Duplicate customer IDs:",
    df_raw["customerID"].duplicated().sum()
)

print(
    "Unique customer IDs:",
    df_raw["customerID"].nunique()
)
```

```python
print("Churn distribution:")
display(df_raw["Churn"].value_counts())

print("\nChurn percentage:")
display(
    df_raw["Churn"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```
